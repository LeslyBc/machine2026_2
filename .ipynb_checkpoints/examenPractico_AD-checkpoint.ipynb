#### 0) Cargar una de las imágenes histológicas

```python
# Utilizar la librería skimage.io para leer la imagen 'histo_x.jpg' en formato RGB.
from skimage import io
import matplotlib.pyplot as plt
import numpy as np
from skimage.filters import gaussian, threshold_otsu 
from skimage.morphology import remove_small_objects 
from skimage.segmentation import flood
from skimage.measure import find_contours 
from skimage.measure import label, regionprops
from scipy.ndimage import binary_fill_holes  

img1 = io.imread('histo_1.jpg')
img2 = io.imread('histo_2.jpg')
```

```python
# Normalizar la imagen para que los píxeles se encuentren en el rango [0, 1]
imagen1normalizada = img1 / 255.0
imagen2normalizada = img2 / 255.0
```

```python
# Visualizar la imagen
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(imagen1normalizada)
plt.title("Célula sin cancer:")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(imagen2normalizada)
plt.title("Célula con cancer:")
plt.axis('off')

plt.show()
```

#### 1) Realizar una transformación de color para convertir la imagen al espacio de color CMYK

```python
# Extraer la componente magenta de la imagen (que corresponde a la región tisular)

# Transformamos RGB a CMYK

# Célula sin cancer
R1 = imagen1normalizada[:,:,0]
G1 = imagen1normalizada[:,:,1]
B1 = imagen1normalizada[:,:,2]

K1 = 1 - np.max(imagen1normalizada, axis=2)
epsilon = 1e-8

M1 = (1 - G1 - K1) / (1 - K1 + epsilon)
M1 = np.clip(M1, 0, 1)

# Célula con cancer
R2 = imagen2normalizada[:,:,0]
G2 = imagen2normalizada[:,:,1]
B2 = imagen2normalizada[:,:,2]

K2 = 1 - np.max(imagen2normalizada, axis=2)

M2 = (1 - G2 - K2) / (1 - K2 + epsilon)
M2 = np.clip(M2, 0, 1)
```

```python
# Visualizar la imagen del canal magenta
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(M1, cmap='gray')
plt.title("No cancer")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(M2, cmap='gray')
plt.title("Cancer")
plt.axis('off')

plt.show()
```

#### 2) Umbralizar la imagen para separar los píxeles del fondo de la región tisular

```python
# Aplicar un filtro gaussiano de tamaño 5x5 y después utilizar el método de Otsu de manera que
# los píxeles correspondientes al lumen y al background de la imagen sean 1s y el resto de los píxeles tengan un valor de 0.
# Nota: Recordar que el método de Otsu requiere como input una imagen en el rango [0-255] en formato "uint8".

# Imagen sin cancer
M1_smooth = gaussian(M1, sigma=1)

M1_uint8 = (M1_smooth * 255).astype(np.uint8)

th1 = threshold_otsu(M1_uint8)

mascara1 = M1_uint8 < th1

# Imagen con cancer
M2_smooth = gaussian(M2, sigma=1)

M2_uint8 = (M2_smooth * 255).astype(np.uint8)

th2 = threshold_otsu(M2_uint8)

mascara2 = M2_uint8 < th2
```

```python
# Visualizar la máscara resultante
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(mascara1, cmap='gray')
plt.title("Mascara sin cancer")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(mascara2, cmap='gray')
plt.title("Mascara con cancer")
plt.axis('off')

plt.show()
```

#### 3) Limpiar la imagen eliminando los artefactos de lumen (objetos blancos pequeños que no son lúmenes)

```python
# Utilizar la librería skimage.morphology.remove_small_objects para eliminar aquellos objetos cuya área sea menor a 300 píxeles
# Más información en https://scikit-image.org/docs/dev/api/skimage.morphology.html#skimage.morphology.remove_small_objects

# Imagen sin cancer
mascara1Limpia = remove_small_objects(mascara1.astype(bool), max_size=300)

# Imagen con cancer
mascara2Limpia = remove_small_objects(mascara2.astype(bool), max_size=300)
```

```python
# Visualizaer la máscara resultante
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(mascara1Limpia, cmap='gray')
plt.title("Mascara limpia sin cancer")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(mascara2Limpia, cmap='gray')
plt.title("Mascara limpia con cancer")
plt.axis('off')

plt.show()
```

#### 4) Rellenar con 0s el fondo de la imagen para quedarnos únicamente con los lúmenes

```python
# Aplicar el algoritmo de expansión a partir de semillas (region growing) de manera que únicamente los lúmenes sean blancos
# y el resto de la imagen negra. Pista: utilizar dos semillas. Nota: Se pueden fijar las semillas de manera manual, pero
# se valorará positivamente a aquell@s que desarrollen una función para encontrarlas automáticamente.
def obtener_lumen(mascara):
    mascara = mascara.astype(bool)
    
    # Crear máscara de fondo que inicialmente será vacía
    fondo = np.zeros_like(mascara, dtype=bool)
    
    # Semillas automáticas en esquinas, con esto aseguramos el fondo
    seeds = [(0,0), (0,-1), (-1,0), (-1,-1)]
    
    for seed in seeds:
        fondo = fondo | flood(mascara, seed_point=seed, connectivity=1)
    
    # Con esto eliminamos el fondo y nos quedamos únicamente con los lúmenes 
    lumen = mascara & (~fondo)
    
    return lumen

lumen1 = obtener_lumen(mascara1Limpia)
lumen2 = obtener_lumen(mascara2Limpia)
```

```python
# Visualizar la máscara resultante. 
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(lumen1, cmap='gray')
plt.title("Lumenes imagen sin cancer")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(lumen2, cmap='gray')
plt.title("Lumenes imagen con cancer")
plt.axis('off')

plt.show()
```

#### 5) Rellenar los objetos de los lúmenes

```python
# Rellenar los lúmenes con la función binary_fill_holes de la librería scipy.ndimage.morphology
lumen1Lleno = binary_fill_holes(lumen1)
lumen2Lleno = binary_fill_holes(lumen2)
```

```python
# Visualizar la máscara resultante
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(lumen1Lleno, cmap='gray')
plt.title("Lúmenes rellenos - Imagen sin cancer")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(lumen2Lleno, cmap='gray')
plt.title("Lúmenes rellenos - Imagen con cancer")
plt.axis('off')

plt.show()
```

#### 6) Detectar y dibujar los contornos de los lúmenes sobre la imagen original

```python
# Dibujar los contornos de los lúmenes en color verde sobre la imagen original RGB. Nota: Utilizar los flags necesarios
# para que los contornos en verde sean perfectamente visibles. 

# Creamos una función para los contornos de la img

def dibujar_los_contornos(image, mask):
    img_out = image.copy()

    # Encontrar contornos
    contours = find_contours(mask, level=0.5)

    for contour in contours:
        contour = contour.astype(int)
        
        # Dibujar en verde (R=0, G=1, B=0)
        img_out[contour[:,0], contour[:,1]] = [0, 1, 0]

    return img_out

# Aplicamos la función para ambas imágenes
imagen1_contornos = dibujar_los_contornos(imagen1normalizada, lumen1Lleno)
imagen2_contornos = dibujar_los_contornos(imagen2normalizada, lumen2Lleno)
```

```python
# Visualizar la imagen superpuesta
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(imagen1_contornos)
plt.title("Contornos imagen sin cancer")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(imagen2_contornos)
plt.title("Contornos imagen con cancer")
plt.axis('off')

plt.show()
```

#### 7) Identificar y cropear el lumen más grande

```python
# Determinar cuál es el lumen de mayor área y hacer un crop del mismo sobre la imagen original RGB.

# Definimos la funcion para ver el lumen mas fuerte

def obtener_recorte_lumen_alto(image, mascara):
    
    # Etiquetamos y obtenemos componentes y propiedades de la imagen
    labeled = label(mascara)
    regions = regionprops(labeled)
    
    # Encontramos el de mayor area
    region_larga = max(regions, key=lambda r: r.area)
    
    # Bounding box
    minr, minc, maxr, maxc = region_larga.bbox
    
    # 1.5) Crop sobre imagen original
    crop = image[minr:maxr, minc:maxc]
    
    return crop

# Aplicamos a las imágenes
crop1 = obtener_recorte_lumen_alto(imagen1normalizada, lumen1Lleno)
crop2 = obtener_recorte_lumen_alto(imagen2normalizada, lumen2Lleno)
```

```python
# Visualizar el lumen cropeado.
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(crop1)
plt.title("Lumen mas grande imagen sin cancer")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(crop2)
plt.title("Lumen mas grande imagen con cancer")
plt.axis('off')

plt.show()
```

#### 8) Extraer 13 características geométricas que permitan caracterizar el lumen recortado 

```python
# Calcular las siguientes características del crop del lumen de mayor área, redondeando su valor hasta el cuarto decimal.
def caracteristicas(mascara):
    
    labeled = label(mascara)
    regions = regionprops(labeled)
    
    # Vemos el lumen más grande
    region = max(regions, key=lambda r: r.area)
    
# 1) Área
    area = region.area
# 2) Área de la bounding box
    bbox_area = region.area_bbox
# 3) Área convexa
    convex_area = region.area_convex
# 4) Exentricidad
    eccentricity = region.eccentricity
# 5) Diámetro equivalente
    equiv_diameter = region.equivalent_diameter_area
# 6) Extensión
    extent = region.extent
# 8) Longitud del eje mayor
    major_axis = region.axis_major_length
# 9) Longitud del eje menor
    minor_axis = region.axis_minor_length
# 10) Orientación
    orientation = region.orientation
# 11) Perímetro
    perimeter = region.perimeter
# 12) Solidez
    solidity = region.solidity
# 13) Compacidad
    compactness = (perimeter**2) / (4 * np.pi * area)

# Lo pongo abajo porque arriba ordenado no valia :p    
# 7) Diámetro Feret
    feret_diameter = major_axis

    redondeo = {
        round(area, 3),
        round(bbox_area, 3),
        round(convex_area, 3),
        round(eccentricity, 3),
        round(equiv_diameter, 3),
        round(feret_diameter, 3),
        round(major_axis, 3),
        round(minor_axis, 3),
        round(orientation, 3),
        round(perimeter, 3),
        round(solidity, 3),
        round(compactness, 3)
    }

    return redondeo
```

```python
redondeo1 = caracteristicas(lumen1Lleno)
redondeo2 = caracteristicas(lumen2Lleno)

print("Caracteristicas sin cancer\n", redondeo1)

print("Caracteristicas con cancer\n", redondeo2)
```
